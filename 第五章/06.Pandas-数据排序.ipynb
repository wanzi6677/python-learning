```python
import pandas as pd

df = pd.read_csv('resours/sales.csv',nrows=20)

#排序
#2.1 根据 销售数量 倒序排序
df.sort_values('销售数量',ascending=False)
#2.2 根据 价格 升序排序
df.sort_values('单价',ascending=True)
#2.3 根据 价格 降序排序，价格一样，再根据 销售数量 升序排序
df.sort_values(['单价','销售数量'],ascending=[False,True])
```
