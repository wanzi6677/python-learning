```python
import pandas as pd

#构造DataFrame --创建数据集
df = pd.DataFrame([
    {"姓名":"张三","语文":90,"数学":80,"英语":75},
    {"姓名":"李四","语文":80,"数学":90,"英语":85},
    {"姓名":"王五","语文":75,"数学":85,"英语":90},
    {"姓名":"赵六","语文":60,"数学":70,"英语":80},
    {"姓名":"孙七","语文":50,"数学":60,"英语":70},
    {"姓名":"周八","语文":91,"数学":50,"英语":78}
])

#统计计算"
print(f"最高分语文:{df['语文'].max()}, 最低分语文:{df['语文'].min()}, 平均分语文:{df['语文'].mean():.2f}")
print(f"最高分数学:{df['数学'].max()}, 最低分数学:{df['数学'].min()}, 平均分数学:{df['数学'].mean():.2f}")
print(f"最高分英语:{df['英语'].max()}, 最低分英语:{df['英语'].min()}, 平均分英语:{df['英语'].mean():.2f}")
```
