```python
import pandas as pd
df = pd.read_csv('resours/sales.csv',nrows=20)
df

#根据产品类别分组，统计各个类别的订单数量-count
df.groupby('产品类别')['订单号'].count()

#根据产品类别分组，统计各个类别的销售数量之和
df.groupby('产品类别')['销售数量'].sum()

#根据产品类别分组，统计各个类别的销售金额之和
df['销售金额'] = df['销售数量']*df['单价']
df.groupby('产品类别')['销售金额'].sum()

#根据产品类别分组，统计各个类别的最低商品单价
df.groupby('产品类别')['单价'].min()

#根据产品类别分组，统计各个类别的平均商品单价
df.groupby('产品类别')['单价'].mean().round(2)

#根据产品类别分组，统计各个类别的平均商品单价,最高单价和平均单价-agg
(df.groupby('产品类别')['单价'].agg(['min','max','mean']))

#根据产品类别分组，统计各个类别的销售数量之和，销售金额之和，平均单价
df.groupby('产品类别').agg({'销售数量':'sum','销售金额':'sum','单价':'mean'})
```
