```python
df3 = pd.read_csv('resours/sales.csv')
df3

#配置项，展示所有的数据
#pd.set_option('display.max_rows', None)

#数据过滤
#1. 获取销售数量 >=10的订单数据
df3[df3['销售数量']>= 10]
#2. 获取产品类别为食品或图书的订单数据
df3[df3['产品类别'].isin(['食品','图书'])]
#3. 获取单价在100-200之间的订单数据,inclusive，可以觉得是否包含边界
df3[df3['单价'].between(100,200,inclusive="left")]
#4. 获取销售数量 >=8并且单价>=100的订单数据
df3[(df3['单价']>=100) & (df3['销售数量']>=8)]
#5. 获取产品类别为服装/食品，支付方式为支付宝/微信支付的订单数据
df3[(df3['产品类别'].isin(['服装','食品'])) & (df3['支付方式'].isin(['微信支付','支付宝']))]
```
