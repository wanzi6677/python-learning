```python
from matplotlib.axes import Axes
import matplotlib.pyplot as plt

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']

#创建子图
#figure:画布对象 ； axes: 子图数组
figure,axes = plt.subplots(nrows=1,ncols=2,figsize = (20,6),dpi = 100)#nrows：行,ncols:列

#图一：柱状图(世界石油储备)-->bar
countries = ['美国','中国','英国','日本','德国','法国','意大利','俄罗斯','韩国','加拿大']
values = [191,165,54,120,79,65,60,180,40,55]

axes1:Axes = axes[0]
axes1.bar(countries,values,width = 0.7,color = 'g')#柱状图使用bar
axes1.set_title('世界石油储备',fontsize=18)
axes1.set_xlabel('国家',fontsize=20)#x标签
axes1.set_ylabel('石油储备(百万桶)',fontsize=20)#y标签

axes1.grid(linestyle = '--',alpha = 0.3)


#图二：饼状图(世界人口）--->擅长比例分析---pie
countries = ['美国','中国','英国','日本','德国','法国','意大利','俄罗斯','韩国','加拿大']
population = [32.9,141,54,120,79,65,60,180,40,55]#人口单位：百万

axes2:Axes = axes[1]
axes2.pie(population,labels = countries,autopct = '%1.1f%%',startangle = 200)
axes2.axis('equal')
axes2.set_title('世界人口比例(单位(百万))',fontsize=18)
axes2.legend(loc = 'lower center',ncol = 5,bbox_to_anchor = (0.5,-0.15))

#保存图片
plt.savefig('resours/01.png')
```
